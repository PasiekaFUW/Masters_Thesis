# Code organisation

Single file with a computer code should not be bigger than $\cal{O}(100)$ lines. The code should be split into 
clear logic blocks using functions and classes (in object oriented languages). Declarations and definitions of 
those units should be separated from a code using them. Separation should be achieved by means of separate files.

In the High Energy Physics (HEP) community we mainly use two languages: 
* ``C++`` - for large experimental frameworks, 
* `Python` - for lighter analysis frameworks. 

The two languages differ very much in the way how many files are combined into a single project.

### C++

Since the C++ is a compiled language more effort is needed to import code located in an external package, or even a separate file.
In general one has to provide paths to all the files contributing to the application and compile those together. 
A simplest way is to list all files in the compilation command:
```Bash
g++ -Wall            # enable all errors warnings
-I./                 # path where compiler looks for header files
ClassA.cxx main.cxx  # list of input implementation files
-o myApplication     # name of the output application file
```

## Version control

Large programming projects are developed with aid of tools for file version control. One of most popular system is 
[git](https://git-scm.com/) created for Linux kernel development. Git is a distributed version control system.
`Distributed` means each developer can work independently on his version of the code. The modifications from each author can be merged
and stored in any place. Usually a dedicated serves - `hub` is used. Two most popular hubs 
are [github](https://github.com/)  and [gitlab](https://about.gitlab.com/). 
Gitlab server can be hosted by anyone, e.g. CERN: [gitlab.cern.ch](http://gitlab.cern.ch/)

Git is convenient way os sharing code between developers and users.
