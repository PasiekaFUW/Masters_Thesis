# Containers

During the class we will use multiple packages and programs specific for High Energy Physics (HEP). To avoid the burden of installing and configuring them on every machine, we will use virtual environments known as containers. Containers are lightweight versions of old time virtual machines. Two popular environments used to run the containers are:
hep
* [Apptainer](https://apptainer.org/) (previously singularity)
* [Docker](https://www.docker.com/)

Both Docker and Apptainer can use the same input files. Docker is available for any operating system, Apptainer only for Linux ones. We will use a
[akalinow/root-fedora39](https://hub.docker.com/repository/docker/akalinow/root-fedora39/general) container prepared for the classes. This is a container based on Fedora Linux distribution, and contains all packages we will use during the classes.

## Remote Access to Computational Machine
Very often, **basically always**, we don't have direct access to the machine where computations are performed. Usually, the user has to take a few steps to have similar functionality:

1) Connect to the remote machine.
2) Start a Jupyter server-type service.
3) Connect to the service from the local computer - terminal.

## Connecting to the Remote Machine

In any respectable computer center, devices in the internal computer network (Local Area Network, LAN) are isolated from the wide network (Wide Area Network, WAN), i.e., the entire internet.

Access to the LAN is obtained through a gateway computer. For students of the Faculty of Physics, this is the
```Bash
tempac.okwf.fuw.edu.pl
```

After logging into the gateway, you can further log in to the machine that provides computational power, for example
computers in the laboratory:

```Bash
pracownia.okwf.fuw.edu.pl
```


**Please:**

* log in directly to the `pracownia` node - it should not work from the FUW.open network if you are using a laptop
* log in first to the `tempac`  and then to the `pracownia`

Instructions:

* **Linux** - open a terminal window.
* **Windows** - applications -> search -> cmd.
In the terminal window, use the ssh command:

```Bash
ssh login@tempac.okwf.fuw.edu.pl
```

After logging in to `tempac`, use the following command:

```Bash
ssh login@pracownia.okwf.fuw.edu.pl
```

After logging in, you can start the relevant service. Of course, logging in two steps is very inconvenient. You can simplify it by using network tunneling. To do this, you have to add an appropriate entry to the configuration of the ssh program (the same location on Linux and Windows):
```Bash
.ssh/config
```

If there is no such file in the user's directory, create one. In the config file, you can put blocks configuring access to remote machines.
In our case, such a block should look like this:

```Bash
Host pracownia
     ForwardX11 yes
     ForwardAgent yes
     UserKnownHostsFile ~/.ssh/known_hosts
     Hostname pracownia.okwf.fuw.edu.pl
     LocalForward 8888 localhost:8888
     RequestTTY yes
     ProxyJump YOUR_LOGIN_ON_TEMPAC@tempac.okwf.fuw.edu.pl:22
     #RemoteCommand ./runSingularity.sh
     User YOUR_LOGIN_ON_TEMPAC
 ```    

 You can set up connection, so some script will be run automatically after login. The following line:
 ```Bash
 RemoteCommand ./runSingularity.sh
 ```

 Will set up execution of the `runSingularity.sh` script after login.

**Please:**

* modify your .ssh/config file and add an entry for the `pracownia` computer.
* log in to the `pracownia` computer:
  ```Bash
  
  ssh pracownia
  ```

## Starting the container on private machine

Following steps assume you have docker desktop aqpplication installed.

**Windows**

* prepare working directory `Docker` and start the container from command line:
```Shell

mkdir %USERPROFILE%\Docker
set WORKING_DIR=%USERPROFILE%\Docker
docker pull akalinow/root-fedora39
docker run --name cepp_lecture -v %WORKING_DIR%:/scratch -p 8888:8888 akalinow/root-fedora39
jupyter lab --no-browser --ip=0.0.0.0 --notebook-dir=$HOME --port 8888
```

**Linux/MacOS**

* Docker version:

```Shell
sudo docker run --name cepp_lecture --rm -e DISPLAY=$DISPLAY  -v /home/$USER:/scratch -v /tmp/.X11-unix:/tmp/.X11-unix:ro -p 8888:8888 akalinow/root-fedora39 
jupyter lab --no-browser --ip=0.0.0.0 --notebook-dir=$HOME --port 8888
```

**Note:**
the Docker container should be started from command line only once. Later you restart it from the Docker desktop application.

* apptainer/singularity version:
  
```Shell
apptainer docker://akalinow/root-fedora39
jupyter lab --no-browser --ip=0.0.0.0 --notebook-dir=$HOME --port 8888
```
